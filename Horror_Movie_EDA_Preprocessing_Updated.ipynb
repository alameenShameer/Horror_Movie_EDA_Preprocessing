## 5. Data Cleaning
**Owner: Nikhil**

### 5.1 Missing Value Identification
Convert whitespace-only strings to missing values and report missingness before deciding how to handle each field.

```python
df_clean = df.copy()

# Standardize text whitespace and convert empty strings to missing values.
text_cols = df_clean.select_dtypes(include="object").columns
for col in text_cols:
    df_clean[col] = df_clean[col].str.strip()
    df_clean[col] = df_clean[col].replace("", pd.NA)

missing_report = pd.DataFrame({
    "missing_count": df_clean.isna().sum(),
    "missing_percent": (df_clean.isna().mean() * 100).round(2)
}).sort_values("missing_count", ascending=False)
display(missing_report)
```

### 5.2 Duplicate Value Removal
Remove exact duplicate rows. Keep the first occurrence. Do not deduplicate using only a movie title because different films can share titles or have different releases.

```python
rows_before_deduplication = len(df_clean)
duplicates_before = df_clean.duplicated().sum()
df_clean = df_clean.drop_duplicates().reset_index(drop=True)
duplicates_after = df_clean.duplicated().sum()

print("Duplicate rows before:", duplicates_before)
print("Rows removed:", rows_before_deduplication - len(df_clean))
print("Duplicate rows after:", duplicates_after)
print("Shape after exact duplicate removal:", df_clean.shape)
```

### 5.3 Handle Inconsistent Data

Normalize text casing/spacing for selected categorical fields. Do not alter descriptive text or movie titles beyond trimming whitespace. Financial zeros are converted to missing because the dataset has a very high proportion of zero budgets/revenues, suggesting that zero often means not reported.

```python
for col in ["original_language", "status"]:
    df_clean[col] = df_clean[col].astype("string").str.strip().str.lower()

# Treat zero budget/revenue/runtime as unavailable values for preprocessing.
# Keep a copy of the raw dataset if zero itself is meaningful for a later analysis.
for col in ["budget", "revenue", "runtime"]:
    df_clean.loc[df_clean[col] == 0, col] = pd.NA

print("Zero placeholders converted to missing values:")
display(df_clean[["budget", "revenue", "runtime"]].isna().sum().to_frame("missing_after_zero_review"))
```

### 5.4 Data Type Correction

Parse `release_date` as a datetime field. Convert `adult` to a categorical/bool-like field and ensure numeric metadata remains numeric.

```python
df_clean["release_date"] = pd.to_datetime(df_clean["release_date"], errors="coerce")

numeric_cols_expected = [
    "popularity", "vote_count", "vote_average",
    "budget", "revenue", "runtime", "collection"
]
for col in numeric_cols_expected:
    df_clean[col] = pd.to_numeric(df_clean[col], errors="coerce")

df_clean["adult"] = df_clean["adult"].astype("boolean")

print(df_clean.dtypes)
```

## 6. Feature Engineering
**Owner: Nikhil**

Create interpretable features from existing columns:
- `release_year` and `release_month` from `release_date`.
- `genre_count` from the comma-separated genre list.
- `primary_genre` from the first listed genre.

The engineered features are derived from existing metadata. The original text and date fields will be excluded from the model feature matrix where appropriate.

```python
df_clean["release_year"] = df_clean["release_date"].dt.year
df_clean["release_month"] = df_clean["release_date"].dt.month

genre_text = df_clean["genre_names"].fillna("")
df_clean["genre_count"] = genre_text.apply(
    lambda value: len([g.strip() for g in value.split(",") if g.strip()])
)
df_clean["primary_genre"] = genre_text.apply(
    lambda value: value.split(",")[0].strip().lower() if value.strip() else pd.NA
)

display(df_clean[[
    "release_date", "release_year", "release_month",
    "genre_names", "genre_count", "primary_genre"
]].head(10))
```

## 9. Target Review and Target Separation
**Owner: Devika**

The target is `vote_average`, a numeric rating on a 0–10 scale, so this is a **regression** task.

In this dataset, a target value of 0 commonly indicates that no audience rating has been recorded rather than a genuine zero-star rating. We inspect its relationship with `vote_count` and exclude zero, missing, and out-of-range targets from supervised training. The original `df_clean` remains unchanged for EDA and auditing.

We create `df_model` and `y` here. The predictor matrix `X_raw` is created only once in Section 10 after the leakage review.

```python
TARGET_COLUMN = "vote_average"

# Ensure the target and vote_count are numeric before reviewing target validity.
df_clean[TARGET_COLUMN] = pd.to_numeric(df_clean[TARGET_COLUMN], errors="coerce")
df_clean["vote_count"] = pd.to_numeric(df_clean["vote_count"], errors="coerce")

print("Target:", TARGET_COLUMN)
print("Target summary before filtering:")
display(df_clean[TARGET_COLUMN].describe())

zero_target_mask = df_clean[TARGET_COLUMN].eq(0)
missing_target_mask = df_clean[TARGET_COLUMN].isna()
invalid_range_mask = df_clean[TARGET_COLUMN].lt(0) | df_clean[TARGET_COLUMN].gt(10)

print("Zero target rows:", int(zero_target_mask.sum()))
print("Missing target rows:", int(missing_target_mask.sum()))
print("Out-of-range target rows:", int(invalid_range_mask.sum()))

print("\nVote count among zero-rated rows:")
display(df_clean.loc[zero_target_mask, "vote_count"].describe())

print("\nZero rating vs. zero vote count:")
display(pd.crosstab(
    zero_target_mask,
    df_clean["vote_count"].eq(0),
    rownames=["vote_average_is_zero"],
    colnames=["vote_count_is_zero"]
))

# Treat zero ratings as unrecorded ratings for this dataset.
# Keep only valid, observed targets in the supervised modelling dataset.
valid_target_mask = (
    df_clean[TARGET_COLUMN].notna()
    & df_clean[TARGET_COLUMN].gt(0)
    & df_clean[TARGET_COLUMN].le(10)
)
df_model = df_clean.loc[valid_target_mask].copy()
y = df_model[TARGET_COLUMN].copy()

print("\nRows before target filtering:", len(df_clean))
print("Rows retained for modelling:", len(df_model))
print("Rows excluded due to unavailable/invalid target:", len(df_clean) - len(df_model))
print("Target shape:", y.shape)
print("Target range after filtering:", y.min(), "to", y.max())
```

## 10. Feature Separation and Data Leakage Review
**Owner: Devika**

Define the prediction scenario before choosing predictors. This notebook uses a **pre-release prediction scenario**: estimate a movie's eventual audience rating using information that could reasonably be available before release.

Therefore, exclude post-release information such as `vote_count`, `popularity`, `revenue`, and release `status`. These can reveal information generated after audiences have watched and rated the movie.

Also exclude identifiers, raw text that is not vectorized, image paths, and original columns already represented by engineered features. Keep `release_year`, `release_month`, `genre_count`, and `primary_genre`.

Create `X_raw` once here. Section 9 creates only the valid modelling rows and target.

```python
# Columns that must not be used as predictors for the pre-release prediction scenario.
drop_from_features = [
    TARGET_COLUMN,       # target: direct target leakage
    "id",                # database identifier
    "original_title",    # high-cardinality title text
    "title",             # title text, not vectorized
    "overview",          # free text, not vectorized in this workflow
    "tagline",           # free text, not vectorized in this workflow
    "release_date",      # represented by release_year and release_month
    "poster_path",       # image asset path, not image features
    "backdrop_path",     # image asset path, not image features
    "genre_names",       # represented by genre_count and primary_genre
    "collection",        # collection identifier, not a numeric magnitude
    "collection_name",   # high-cardinality collection label
    "adult",             # constant in this dataset
    "vote_count",        # post-release audience activity
    "popularity",        # post-release/dynamic engagement measure
    "revenue",            # post-release outcome
    "status"              # may reveal whether the movie has already been released
]

X_raw = df_model.drop(
    columns=[col for col in drop_from_features if col in df_model.columns]
).copy()

removed_columns = [col for col in drop_from_features if col in df_model.columns]

print("Columns removed from predictors:")
print(removed_columns)
print("\nRemaining predictor columns:")
print(X_raw.columns.tolist())
print("\nFeature shape:", X_raw.shape)
print("Target shape:", y.shape)
print("Target accidentally present in X:", TARGET_COLUMN in X_raw.columns)

assert TARGET_COLUMN not in X_raw.columns
assert X_raw.index.equals(y.index)
```

## 11. Train-Test Splitting
**Owner: Vyas**

Split the predictor rows and continuous target together so that each feature row stays paired with its correct `vote_average`. All learned feature-preprocessing decisions (imputation, outlier bounds, scaling and encoding) are fitted on `X_train` and then applied to `X_test`. The target remains on its original scale.


```python
X_train, X_test, y_train, y_test = train_test_split(
    X_raw,
    y,
    test_size=0.20,
    random_state=42
)

print("Training feature shape:", X_train.shape)
print("Testing feature shape:", X_test.shape)
print("Training target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)
print("\nTraining target summary:")
print(y_train.describe().to_frame("vote_average"))
print("\nTesting target summary:")
print(y_test.describe().to_frame("vote_average"))
```

## 12. Numerical Data Preprocessing
**Owner: Vyas**

### 12.1 Separate Numerical Columns
Separate numeric and categorical columns based on the training partition. Datetime columns have already been converted into numeric year/month features.

```python
numeric_features = X_train.select_dtypes(include=["number"]).columns.tolist()
categorical_features = X_train.select_dtypes(include=["object", "string", "category", "bool", "boolean"]).columns.tolist()

print("Numerical columns:", numeric_features)
print("Categorical columns:", categorical_features)
```

### 12.2 Numerical Missing Value Handling

Use median imputation because several numeric variables are skewed and contain extreme values. Fit the imputer on training data only.

```python
numeric_imputer = SimpleImputer(strategy="median")
X_train_num = pd.DataFrame(
    numeric_imputer.fit_transform(X_train[numeric_features]),
    columns=numeric_features,
    index=X_train.index
)
X_test_num = pd.DataFrame(
    numeric_imputer.transform(X_test[numeric_features]),
    columns=numeric_features,
    index=X_test.index
)
print("Missing numerical values after imputation (train):", X_train_num.isna().sum().sum())
print("Missing numerical values after imputation (test):", X_test_num.isna().sum().sum())
```

### 12.3 Outlier Detection and Treatment

Use the IQR rule to **identify** potential outliers using training data only. An IQR flag does not automatically mean a value is an error: very high budgets or long runtimes may be genuine movie characteristics.

Do not cap every numerical column blindly. Review the summary and plots first. Add only columns that have a defensible capping decision to `OUTLIER_TREATMENT_COLUMNS`. Any selected bounds are learned from training data and then applied unchanged to the test data.

```python
outlier_summary = []
outlier_bounds = {}

for col in numeric_features:
    q1 = X_train_num[col].quantile(0.25)
    q3 = X_train_num[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    outlier_count = ((X_train_num[col] < lower) | (X_train_num[col] > upper)).sum()

    outlier_bounds[col] = (lower, upper)
    outlier_summary.append({
        "feature": col,
        "lower_bound": lower,
        "upper_bound": upper,
        "outlier_count": int(outlier_count),
        "outlier_percent": round(outlier_count / len(X_train_num) * 100, 2)
    })

outlier_report = pd.DataFrame(outlier_summary).sort_values(
    "outlier_percent", ascending=False
)
display(outlier_report)

# Select columns for capping only after reviewing the report and domain meaning.
# Leave empty by default so genuine extreme movie values are not altered blindly.
OUTLIER_TREATMENT_COLUMNS = []

for col in OUTLIER_TREATMENT_COLUMNS:
    if col not in outlier_bounds:
        raise ValueError(f"{col} is not a numerical training feature.")

    lower, upper = outlier_bounds[col]
    X_train_num[col] = X_train_num[col].clip(lower=lower, upper=upper)
    X_test_num[col] = X_test_num[col].clip(lower=lower, upper=upper)

print("Columns capped:", OUTLIER_TREATMENT_COLUMNS)
print("Bounds were calculated from training data only.")
```

### 12.4 Feature Scaling

`StandardScaler` is the primary scaling method. Fit scaler on training data only.

```python
standard_scaler = StandardScaler()
X_train_num_scaled = pd.DataFrame(
    standard_scaler.fit_transform(X_train_num),
    columns=numeric_features,
    index=X_train_num.index
)
X_test_num_scaled = pd.DataFrame(
    standard_scaler.transform(X_test_num),
    columns=numeric_features,
    index=X_test_num.index
)
```

## 13. Categorical Data Preprocessing
**Owner: Al Ameen**

### 13.1 Separate Categorical Columns

The categorical feature list was created in Step 12 from the training partition.

```python
print("Categorical columns to preprocess:", categorical_features)
```

### 13.2 Categorical Missing Value Handling

Fill missing categorical values with the most frequent training category. The imputer is fitted only on training data.

```python
if categorical_features:
    categorical_imputer = SimpleImputer(strategy="most_frequent")
    X_train_cat_imputed = pd.DataFrame(
        categorical_imputer.fit_transform(X_train[categorical_features]),
        columns=categorical_features,
        index=X_train.index
    )
    X_test_cat_imputed = pd.DataFrame(
        categorical_imputer.transform(X_test[categorical_features]),
        columns=categorical_features,
        index=X_test.index
    )
else:
    X_train_cat_imputed = pd.DataFrame(index=X_train.index)
    X_test_cat_imputed = pd.DataFrame(index=X_test.index)

print("Missing categorical values after imputation (train):", X_train_cat_imputed.isna().sum().sum())
```

### 13.3 One-Hot Encoding

Use one-hot encoding for nominal categories such as language, status and primary genre. `handle_unknown='ignore'` allows categories found only in the test set without failing transformation.

```python
if categorical_features:
    try:
        onehot_encoder = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
    except TypeError:  # compatibility with older scikit-learn versions
        onehot_encoder = OneHotEncoder(handle_unknown="ignore", sparse=False)

    train_cat_array = onehot_encoder.fit_transform(X_train_cat_imputed)
    test_cat_array = onehot_encoder.transform(X_test_cat_imputed)
    encoded_feature_names = onehot_encoder.get_feature_names_out(categorical_features)

    X_train_cat_encoded = pd.DataFrame(
        train_cat_array, columns=encoded_feature_names, index=X_train.index
    )
    X_test_cat_encoded = pd.DataFrame(
        test_cat_array, columns=encoded_feature_names, index=X_test.index
    )
else:
    X_train_cat_encoded = pd.DataFrame(index=X_train.index)
    X_test_cat_encoded = pd.DataFrame(index=X_test.index)

print("One-hot encoded training shape:", X_train_cat_encoded.shape)
```

## 14. Concatenate Numerical and Categorical DataFrames
**Owner: Al Ameen**

Combine the scaled numerical columns and one-hot encoded categorical columns. Align rows by their original indices.

```python
X_train_processed = pd.concat([X_train_num_scaled, X_train_cat_encoded], axis=1).sort_index(axis=1)
X_test_processed = pd.concat([X_test_num_scaled, X_test_cat_encoded], axis=1).sort_index(axis=1)

# Ensure test columns match training columns exactly.
X_test_processed = X_test_processed.reindex(columns=X_train_processed.columns, fill_value=0)

print("Processed training shape:", X_train_processed.shape)
print("Processed testing shape:", X_test_processed.shape)
display(X_train_processed.head())
```

## 15.2 Feature-to-Target Correlation

###15.3 Feature Selection
Use VarianceThreshold fit on X_train_processed to drop zero-variance columns, then apply it to X_test_processed.

```python
# Feature Selection
selector = VarianceThreshold(threshold=0.0)
X_train_sel = selector.fit_transform(X_train_processed)
X_test_sel = selector.transform(X_test_processed)

kept_cols = X_train_processed.columns[selector.get_support()]
X_train_final = pd.DataFrame(X_train_sel, columns=kept_cols, index=X_train_processed.index)
X_test_final = pd.DataFrame(X_test_sel, columns=kept_cols, index=X_test_processed.index)

print("Before selection:", X_train_processed.shape[1], "columns")
print("After selection:", X_train_final.shape[1], "columns")
```
